```python
from xbbg import blp
import pandas as pd
import matplotlib.pyplot as plt
```

```python
END = pd.Timestamp.today().normalize()
START = pd.Timestamp("2019-01-01")
```

```python
TICKERS = {
"EURUSD Curncy": "EUR/USD",
"EURGBP Curncy": "EUR/GBP",
"MO1 Comdty": "EUA_Front",
"UKE1 Comdty": "UKA_M1",
"XA1 Comdty": "COAL_M1",
"QS1 Comdty": "LS_Oil_M1",
"TZT1 Comdty": "TTF_M1",
"FN1 Comdty": "NBP_M1",
"FPA1 Comdty": "DE_BASE_M1",
"ELUB1MON BCFV Index": "UK_BASE_M1",
}
```

```python
raw = blp.bdh(
    list(TICKERS),
    flds=["PX_LAST"],
    start_date=pd.Timestamp(START).strftime("%Y-%m-%d"),
    end_date=pd.Timestamp(END).strftime("%Y-%m-%d")
).to_pandas()

df = raw = raw.pivot(index="date", columns="ticker", values="value").rename(columns=TICKERS)
df.index = pd.to_datetime(df.index)
df.to_csv("data.csv", index=True)
```
