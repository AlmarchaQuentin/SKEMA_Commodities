```python
import pandas as pd
lng = pd.read_csv(r"C:\Users\quent\Documents\SKEMA\COURSES\Commodities\projects\3\1. Raw Data\lng_sendout.csv")
```

```python
lng["gasDayStart"] = pd.to_datetime(lng["gasDayStart"])
lng = lng.sort_values("gasDayStart")

lng["lng7"] = lng["sendOut"].rolling(7, min_periods=7).sum() / 1000  # TWh over 7 gas days
lng["lng_dev"] = lng["lng7"] - lng["lng7"].shift(1).rolling(365, min_periods=365).mean()
```

```python
lng = lng.drop(columns=["sendOut", "status"])
```

```python
cols = lng.columns.drop("gasDayStart")
lng[cols] = lng[cols].shift(2)  # gas day D is usable at D+2
```

```python
lng.to_csv("lng_processed.csv", index=False)
```
