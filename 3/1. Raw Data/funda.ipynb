```python
import time
import requests
import pandas as pd
from requests.adapters import HTTPAdapter
from urllib3.util.retry import Retry

API_KEY = "7b9a77a6449ac6e27a6e9b6c54b1d7b6"
URL = "https://agsi.gie.eu/api"

FIELDS = [
    "gasDayStart", "gasInStorage", "workingGasVolume", "full", "trend",
    "injection", "withdrawal", "netWithdrawal", "injectionCapacity",
    "withdrawalCapacity", "consumption", "consumptionFull", "status",
]

# Session with automatic retries on timeouts / server errors
s = requests.Session()
s.headers.update({"x-key": API_KEY})
s.mount("https://", HTTPAdapter(max_retries=Retry(
    total=5, backoff_factor=2, status_forcelist=[429, 500, 502, 503, 504],
    allowed_methods=["GET"],
)))

params = {
    "type": "eu",          # EU aggregate
    "from": "2000-01-01",  # data actually starts ~2011
    "to": "2026-10-10",
    "size": 60,            # smaller pages = faster responses
    "page": 1,
}

rows = []
while True:
    r = s.get(URL, params=params, timeout=120)
    r.raise_for_status()
    js = r.json()
    rows.extend({k: d.get(k) for k in FIELDS} for d in js["data"])  # keep only needed fields
    last = int(js.get("last_page", 1))
    print(f"page {params['page']}/{last}")
    if params["page"] >= last:
        break
    params["page"] += 1
    time.sleep(1)  # stay under the API rate limit

df = pd.DataFrame(rows, columns=FIELDS)

# API returns numbers as strings -> convert
num_cols = [c for c in FIELDS if c not in ("gasDayStart", "status")]
df[num_cols] = df[num_cols].apply(pd.to_numeric, errors="coerce")
df["gasDayStart"] = pd.to_datetime(df["gasDayStart"])
df = df.sort_values("gasDayStart").reset_index(drop=True)

print(df.tail())
df.to_csv("funda.csv", index=False)
```
