# TTF M+1 / M+2 & European gas storage

```python
import warnings
warnings.filterwarnings("ignore")

import pandas as pd
import matplotlib.pyplot as plt
from xbbg import blp
```

```python
END = pd.Timestamp.today().normalize()
START = pd.Timestamp("2011-03-31")  

TICKERS = {
    "TZT1 Comdty": "TTF_M1",
    "TZT2 Comdty": "TTF_M2",
    "TZT3 Comdty": "TTF_M3",
    "XAX6 Comdty": "Coal_M1",
    "EURUSD BGN Curncy": "EUR/USD",
    "MO1 Comdty": "EUA_M1"
}
```

```python
raw = blp.bdh(
    list(TICKERS),
    "PX_LAST",
    start_date=pd.Timestamp(START).strftime("%Y-%m-%d"),
    end_date=pd.Timestamp(END).strftime("%Y-%m-%d"),
).to_pandas()

# Long (ticker / date / value) -> wide, one column per series
df = raw.pivot(index="date", columns="ticker", values="value").rename(columns=TICKERS)
df.index = pd.to_datetime(df.index)
df = df[list(TICKERS.values())]
# df.columns.name = None

df.Coal_M1.plot(title="Coal Futures Price", ylabel="$/t", xlabel="Date", figsize=(12, 6))
```

```python
df["M1_M2_spread"] = df.TTF_M1 - df.TTF_M2
df["M1_M3_spread"] = df.TTF_M1 - df.TTF_M3
df["M2_M3_spread"] = df.TTF_M2 - df.TTF_M3
```
