# Archived weather forecast
The raw seven-day anomaly is retained and becomes available one calendar day after its inferred issue date. Exact run timestamps are unverified. The source contains only a changing-horizon aggregate, so a same-valid-horizon forecast revision cannot be reconstructed and is not invented.

```python
from pathlib import Path
import os
import numpy as np
import pandas as pd

ROOT = Path(os.environ.get("TTF_PROJECT_ROOT", Path.cwd().parent))
RAW = ROOT / "1. Raw Data"
OUT = ROOT / "2. Data Processing"
fc = pd.read_csv(RAW / "weather_forecast.csv", parse_dates=["date"])
fc = fc.sort_values("date").reset_index(drop=True)
fc["fc_source_issue_date"] = fc["date"]
fc["fc_available_date"] = fc["fc_source_issue_date"] + pd.Timedelta(days=1)
fc["fc_run_timestamp_verified"] = False
fc["date"] = fc["fc_available_date"]
fc.to_csv(OUT / "weather_forecast_processed.csv", index=False)
```
